```python
import warnings
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from pathlib import Path
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
                            classification_report,
                            confusion_matrix,
                            accuracy_score,
                            roc_auc_score,
                            roc_curve,
                            precision_recall_curve,
                            f1_score,
                            precision_score,
                            recall_score
                            )
warnings.filterwarnings("ignore")
```

### 1. Setting-up Configurations

```python
SEED = 42
PROCESSED_DATA_PATH = Path('../../data/processed')
FIGURE_DIR = Path('../../artifacts/figures')
```

### 2. Load Data

```python
X_train = np.load(f"{PROCESSED_DATA_PATH}/X_train.npz")['arr_0']
X_test = np.load(f"{PROCESSED_DATA_PATH}/X_test.npz")['arr_0']
Y_train = np.load(f"{PROCESSED_DATA_PATH}/Y_train.npz")['arr_0']
Y_test = np.load(f"{PROCESSED_DATA_PATH}/Y_test.npz")['arr_0']
```

### 3. Logistic Regression 

```python
def sigmoid(z):
    return 1 / (1+np.exp(-z))
```

```python
X_linespace = np.linspace(-10, 10, 100)
Y_hat = sigmoid(X_linespace)

plt.plot(X_linespace, Y_hat)
plt.xlabel("Z")
plt.ylabel("Y_hat")
plt.show()
```

```python
model_lr = LogisticRegression(
                                random_state = 42,
                                max_iter = 1000,
                            )
                            
model_lr.fit(X_train, Y_train)
```

```python
"""Inference model"""

Y_hat_train = model_lr.predict(X_train)
Y_hat_test = model_lr.predict(X_test)

Y_hat_train, Y_hat_test
```

```python
Y_proba_train = model_lr.predict_proba(X_train)
Y_proba_test = model_lr.predict_proba(X_test)

Y_proba_train,  Y_proba_test
```

### 4. Model evaluation

```python
""" Training Data Accuracy """

train_accuracy = accuracy_score(Y_train, Y_hat_train)
train_precision = precision_score(Y_train, Y_hat_train)
train_recall = recall_score(Y_train, Y_hat_train)
train_f1_score = f1_score(Y_train, Y_hat_train)
```

```python
print(f"Accuracy score : {train_accuracy}")
print(f"Precision score : {train_precision}")
print(f"Recall score : {train_recall}")
print(f"f1 score : {train_f1_score}")
```

```python
cm = confusion_matrix(Y_train, Y_hat_train)

sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel("Predicted")
plt.ylabel("Actuat")
plt.title("Confusion matrix")
plt.show()
```

```python
""" Testing Data Accuracy """

test_accuracy = accuracy_score(Y_test, Y_hat_test)
test_precision = precision_score(Y_test, Y_hat_test)
test_recall = recall_score(Y_test, Y_hat_test)
test_f1_score = f1_score(Y_test, Y_hat_test)


print(f"Accuracy score : {test_accuracy}")
print(f"Precision score : {test_precision}")
print(f"Recall score : {test_recall}")
print(f"f1 score : {test_f1_score}")
```

```python
cm = confusion_matrix(Y_test, Y_hat_test)

sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel("Predicted")
plt.ylabel("Actuat")
plt.title("Confusion matrix")
plt.show()
```
